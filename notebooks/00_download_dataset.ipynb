```python
import os
import kagglehub

# Configura la ruta al kaggle.json
os.environ["KAGGLE_CONFIG_DIR"] = "/home/zeus/mnt/code/anomaly_detection/.kaggle"
path = kagglehub.dataset_download("ravikagglex/shanghaitech-anomaly-detection")

print("✅ Dataset descargado en:", path)
```

```python
from pathlib import Path
import shutil

src = Path(path)  # ruta devuelta por kagglehub
dest = Path("/teamspace/studios/this_studio/anomaly_detection/data/raw")
dest.mkdir(parents=True, exist_ok=True)
```

```python
shutil.copytree(src, dest, dirs_exist_ok=True)
print("✅ Datos copiados a:", dest)
```

```python
# --- 2️⃣ Verificación rápida adaptada al nuevo dataset ---
from pathlib import Path
dest = Path("/teamspace/studios/this_studio/anomaly_detection/data/raw/dataset/mp")
# Contar cantidad de videos
videos = [d for d in dest.iterdir() if d.is_dir()]
print(f"🎥 Total de carpetas de video: {len(videos):,}")

# Contar frames y optical flow por tipo
total_frames = sum(1 for _ in dest.rglob("frames/*.jpg"))
total_flow   = sum(1 for _ in dest.rglob("optical_flow/*.jpg"))

print(f"🖼️ Total de frames RGB: {total_frames:,}")
print(f"🌪️ Total de optical flow: {total_flow:,}")
```
